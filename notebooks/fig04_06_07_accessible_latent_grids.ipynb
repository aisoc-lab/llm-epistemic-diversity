# Figures 4, 6, 7: Cumulative gain: accessible vs. latent

```python
import sys
from pathlib import Path

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))

from src.plots.style import apply_paper_style, model_label, model_color, ordered_models
apply_paper_style()
```

```python
import matplotlib.pyplot as plt

from src.data_loader import load_accessible_professions, load_latent_professions, cumulative_unique_by_sample, try_load, warn_incomplete_coverage
from src.plots.cumulative_gain import plot_cumulative_grid, plot_cumulative_single_row
from src.plots.style import PROFESSIONS_ALL, FORMATS_ALL

accessible = try_load(load_accessible_professions, label="accessible professions")
latent = try_load(load_latent_professions, expanded=True, label="latent professions (expanded)")

cum_acc = None
if accessible is not None:
    warn_incomplete_coverage(accessible, label="accessible", col="profession", expected=PROFESSIONS_ALL)
    cum_acc = cumulative_unique_by_sample(accessible, group_cols=["model_version", "profession", "prompt_key"], order_col="sample_id")

cum_lat = None
if latent is not None:
    warn_incomplete_coverage(latent, label="latent", col="profession", expected=PROFESSIONS_ALL)
    # Latent's "expanded" rows are already first-occurrence-deduplicated per query, in name_rank order:
    # passing them through cumulative_unique_by_sample with order_col="name_rank" recovers the exact same
    # cumulative curve (every row is trivially a first occurrence), while keeping one code path for both protocols.
    cum_lat = cumulative_unique_by_sample(latent, group_cols=["model_version", "profession", "prompt_key"], order_col="name_rank")
```

## Fig 4a: Accessible, one row only

```python
if cum_acc is None:
    print("Skipping: no accessible data loaded.")
else:
    fig, axes = plot_cumulative_single_row(cum_acc, "chemist", row_col="profession", col_col="format", col_order=FORMATS_ALL, max_x=100)
    plt.show()
```

## Fig 4b: Latent, one row only

```python
if cum_lat is None:
    print("Skipping Fig 4b: no latent data loaded.")
else:
    fig, axes = plot_cumulative_single_row(cum_lat, "chemist", row_col="profession", col_col="format", col_order=FORMATS_ALL, max_x=100)
    plt.show()
```

## Fig 6: [Accessible] full profession x format grid

```python
if cum_acc is None:
    print("Skipping Fig 6: no accessible data loaded.")
else:
    fig, axes = plot_cumulative_grid(cum_acc, row_col="profession", col_col="format", row_order=PROFESSIONS_ALL, col_order=FORMATS_ALL, max_x=100, log_y=True)
    plt.show()
```

## Fig 7: [Latent] full profession x format grid

```python
if cum_lat is None:
    print("Skipping Fig 7: no latent data loaded.")
else:
    fig, axes = plot_cumulative_grid(cum_lat, row_col="profession", col_col="format", row_order=PROFESSIONS_ALL, col_order=FORMATS_ALL, max_x=100)
    plt.show()
```
