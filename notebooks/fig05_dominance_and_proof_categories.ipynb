# Figure 5: Dominance CDF and proof categories

```python
import sys
from pathlib import Path

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))

from src.plots.style import apply_paper_style, model_label, model_color, ordered_models
apply_paper_style()
```

```python
import matplotlib.pyplot as plt

from src.data_loader import load_accessible_professions, load_proofs, try_load, warn_incomplete_coverage
from src.plots.dominance import top_entity_dominance, plot_dominance_cdf
from src.plots.proof_categories import plot_stacked_proof_categories
from src.plots.style import PROBLEM_IDS_ALL

professions = try_load(load_accessible_professions, label="accessible professions")
proofs = try_load(load_proofs, label="proofs")
if proofs is not None:
    warn_incomplete_coverage(proofs, label="proofs", col="problem_id", expected=PROBLEM_IDS_ALL)
```

## 5a: CDF of top-entity dominance per query

```python
if professions is None:
    print("Skipping Fig 5a: no professions data loaded.")
else:
    shares = top_entity_dominance(professions, valid_only=True)
    fig, ax = plt.subplots(figsize=(6, 5))
    plot_dominance_cdf(shares, ax=ax)
    plt.tight_layout()
    plt.show()
```

## 5b: Stacked proof categories (single problem)

```python
problem_id = "1"

if proofs is None:
    print("Skipping Fig 5b: no proofs data loaded.")
else:
    fig, ax = plt.subplots(figsize=(7, 5))
    plot_stacked_proof_categories(proofs, problem_id=problem_id, ax=ax)
    ax.set_ylabel("# Proofs for Category")
    plt.tight_layout()
    plt.show()
```
