# Table 3: Coverage of Wikipedia reference lists by profession

```python
import sys
from pathlib import Path

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))

from src.plots.style import apply_paper_style, model_label, model_color, ordered_models
apply_paper_style()
```

```python
import os

from src.data_loader import load_accessible_professions, load_reference_names, try_load, warn_incomplete_coverage
from src.metrics.coverage import coverage_by_profession, unique_reference_coverage_pct
from src.plots.style import PROFESSIONS_ALL

accessible = try_load(load_accessible_professions, label="accessible")
if accessible is not None:
    warn_incomplete_coverage(accessible, label="accessible", col="profession", expected=PROFESSIONS_ALL)
```

## Per-response coverage breakdown (found in source / valid-not-in-source / invalid-not-in-source)

```python
per_response = None
if accessible is None:
    print("Skipping: no accessible data loaded.")
else:
    per_response = coverage_by_profession(accessible)
    per_response = per_response.round(1)
per_response
```

## Reference-list coverage

```python
table3 = None
if accessible is None:
    print("Skipping: no accessible data loaded.")
else:
    reference_sizes = {p: len(load_reference_names(p)) for p in accessible["profession"].unique()}
    table3 = unique_reference_coverage_pct(accessible, reference_sizes)
    table3 = table3.round(1)
table3
```

```python
if table3 is None:
    print("Skipping export: Table 3 not computed.")
else:
    out_path = "../data/results/metrics/table3_coverage.csv"
    os.makedirs(os.path.dirname(out_path), exist_ok=True)
    table3.round(2).to_csv(out_path, index=False)
    print("wrote", out_path)
    cov = table3["coverage_pct"]
    print("min={:.1f}%, max={:.1f}%".format(cov.min(), cov.max()))
```
