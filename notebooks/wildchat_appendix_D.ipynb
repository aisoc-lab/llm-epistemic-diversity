```python
import os

from src.data_loader import load_wildchat, try_load

wildchat = try_load(load_wildchat, label="wildchat")
if wildchat is not None:
    print(len(wildchat), "annotated queries")
```

```python
counts = wildchat["number_of_responses"].value_counts() if wildchat is not None else None
counts
```

```python
if wildchat is None:
    print("Skipping: no wildchat data loaded.")
else:
    multi_answer_pct = 100 * (counts.get("multiple", 0) + counts.get("infinite", 0)) / len(wildchat)
    underspecified_pct = 100 * (wildchat["specification"] == "underspecified").mean()
    print("multiple={}, infinite={}".format(counts.get("multiple", 0), counts.get("infinite", 0)))
    print("multi-answer: {:.1f}%".format(multi_answer_pct))
    print("underspecified: {:.1f}%".format(underspecified_pct))
```

```python
if wildchat is None:
    print("Skipping export: no wildchat data loaded.")
    summary = None
else:
    out_path = "../data/results/metrics/wildchat_appendix_D_summary.csv"
    os.makedirs(os.path.dirname(out_path), exist_ok=True)
    summary = wildchat.groupby(["number_of_responses", "specification"]).size().reset_index(name="count")
    summary.to_csv(out_path, index=False)
    print("wrote", out_path)
summary
```
